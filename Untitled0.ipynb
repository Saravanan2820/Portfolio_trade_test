```python

def check_trade(trade):

    # STEP 1: Validate trade data

    quantity = trade['quantity']
    price = trade['price']
    action = trade['action']

    if quantity <= 0 or price <= 0 or action not in ['Buy', 'Sell']:
        return {
            "decision": "Rejected",
            "reason": "Invalid trade data."
        }

    # STEP 2: Restricted security check

    portfolio_id = trade['portfolio_id']
    security_id = trade['security_id']

    portfolio = portfolio_data[portfolio_id]

    if security_id in portfolio['restricted_securities']:
        return {
            "decision": "Rejected",
            "reason": "Security is restricted for this portfolio."
        }

    # STEP 3: Sell holdings check

    if action == 'Sell':

        current_holdings = portfolio['current_holdings']

        held_quantity = current_holdings.get(
            security_id, {}
        ).get('quantity', 0)

        if held_quantity < quantity:
            return {
                "decision": "Rejected",
                "reason": "Insufficient holdings to sell."
            }

    # STEP 4: Buy affordability check

    if action == 'Buy':

        purchase_cost = quantity * price
        cash_balance = portfolio['cash_balance']

        if purchase_cost > cash_balance:
            return {
                "decision": "Rejected",
                "reason": "Insufficient cash for this buy order."
            }

        # STEP 5: Position concentration check

        post_trade_cash = cash_balance - purchase_cost

        existing_quantity = portfolio['current_holdings'].get(
            security_id, {}
        ).get('quantity', 0)

        post_trade_quantity = existing_quantity + quantity

        post_trade_value_of_this_security = (
            post_trade_quantity * price
        )

        value_of_other_holdings = sum(
            holding['quantity'] * holding['market_price']
            for sid, holding in portfolio['current_holdings'].items()
            if sid != security_id
        )

        total_portfolio_value_after = (
            post_trade_cash
            + post_trade_value_of_this_security
            + value_of_other_holdings
        )

        position_pct = (
            post_trade_value_of_this_security
            / total_portfolio_value_after
        )

        if position_pct > portfolio['max_position_pct']:
            return {
                "decision": "Needs Manual Sign-off",
                "reason": "Exceeds maximum position concentration — requires approval."
            }

    return {
        "decision": "Approved",
        "reason": "Trade passed Steps 1 to 5."
    }
```

```python

# STEP 6: Process all trades

results = []

for index, row in df.iterrows():

    trade = row.to_dict()

    result = check_trade(trade)

    results.append(result)

# Create final compliance report

compliance_report = pd.concat(
    [df.reset_index(drop=True), pd.DataFrame(results)],
    axis=1
)

print("COMPLIANCE REPORT")
display(compliance_report)
```

```python

decision_summary = compliance_report['decision'].value_counts()

print("TRADE DECISION SUMMARY")
print(decision_summary)
```
